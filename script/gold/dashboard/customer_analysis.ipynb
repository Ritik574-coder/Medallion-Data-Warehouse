```python
import os
import pandas as pd
from tabulate import tabulate
import plotly.express as px
from sqlalchemy import create_engine
from urllib.parse import quote_plus
```

```python
DB_USER     = os.getenv("DB_USER")
DB_PASSWORD = quote_plus(os.getenv("DB_PASSWORD"))
DB_HOST     = "localhost"
DB_PORT     = "1433"
DB_NAME     = "TestDB"
DRIVER      = ("ODBC Driver 18 for SQL Server")
TRUST_SERVER_CERTIFICATE = "yes"

connection_string = (
    f"mssql+pyodbc://{DB_USER}:{DB_PASSWORD}"
    f"@{DB_HOST}:{DB_PORT}/{DB_NAME}"
    f"?driver={quote_plus(DRIVER)}"
    f"&TrustServerCertificate={TRUST_SERVER_CERTIFICATE}"
)

engine = create_engine(connection_string)
```

```python
query = """
SELECT
    YEAR(account_created_date) AS account_created_year,
    COUNT(*) AS customer_count
FROM gold.dim_customers
GROUP BY YEAR(account_created_date)
ORDER BY account_created_year
"""
```

```python
customer_over_year = pd.read_sql(query , engine)
```

```python
customer_over_year
```

```python


fig = px.bar(
    customer_over_year,
    x="account_created_year",
    y="customer_count",
    text="customer_count",
    title="Customer Acquisition by Year"
)

fig.update_layout(
    title={
        "text": "<b>Customer Acquisition by Year</b>",
        "x": 0.5,          # center
        "xanchor": "center",
        "font": {
            "size": 24
        }
    },
    xaxis_title="<b>Account Created Year</b>",
    yaxis_title="<b>Customer Count</b>",
    template="plotly_white",
    width=1000,
    height=600
)

fig.update_traces(
    textposition="outside"
)

fig.show()
```
